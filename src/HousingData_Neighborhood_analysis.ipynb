```python
# Scaling and dimentionality reduction
```

```python
# Additional Libraries to import
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
housingDataClean = housingData.copy()
```

```python
missingValues = housingDataClean.isnull().sum().sort_values(ascending=False)
```

```python
missingValues[missingValues > 0]
```

```python
threshold = len(housingData)*0.5
# drop those columns which have more than 50% nulls - based on what? how do we decide that?
threshold
```

```python
housingDataClean = housingDataClean.dropna(thresh=threshold, axis=1)
housingDataClean.shape
```

```python
# impute remaining missing values

# creating copy for comparison
hdc = housingDataClean.copy()

# 1. for numeric colums: fill with median
numericCols = hdc.select_dtypes(include=[np.number]).columns
# this will create dataframe numericCols, having only those columns in housingData that have numeric (not alpha, or alphanumeric) data

for col in numericCols:
    hdc[col] = hdc[col].fillna(hdc[col].median())

# 2. For categorical columns: fill with mode
categoricalCols = hdc.select_dtypes(include=['object']).columns

for col in categoricalCols:
    hdc[col] = hdc[col].fillna(housingData[col].mode()[0])

# why is housingData used instead of hdc for finding the mode?
```

```python
hdc.isnull().sum()
```

```python
# one hot encoding - get_dummies
# This data, even though has more columns, has expanded each column into 0s and 1s, something that computer can process quicker. Technically, and obviously, if we compress it, it will shrink to 76 columns.
hdc = pd.get_dummies(hdc, drop_first=True)
hdc.shape
```

```python
# before using PCA (principal component analysis), we need to standardize the data, and hence scale the data that has mean and std deviation of 0 and 1.
# https://share.google/aimode/H2cTDiJgz2lOOeD1D
scaler = StandardScaler()
scaledData = scaler.fit_transform(hdc)
print(type(scaler), type(scaledData), scaledData.shape)
```

```python
pca=PCA()
pca.fit(scaledData)

#plot explained variance
plt.figure(figsize=(10,6))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel("Number of components")
plt.ylabel("Cumulative explained variance")
plt.title("Explained variance by principal components in PCA")
plt.grid(True)
plt.show()
```

```python
pca = PCA(n_components=0.95) #what is n_components?
finalCleanedData = pca.fit_transform(scaledData)
print(f"Original shape: {housingData.shape}")
print(f"Shape after categorical variables and one hot encoding: {scaledData.shape}")
print(f"Reduced shape to 95%: {finalCleanedData.shape}")
```

```python
from sklearn.cluster import KMeans
```

```python
sampleDf = hdc.sample(frac=0.1, random_state=42) #10% sample rate
```

```python
features = ['GrLivArea', 'LotArea', 'OverallQual']
xScaled = scaler.fit_transform(sampleDf[features])
```

```python
import multiprocessing, os
try:
    print (f"Physical Cores: {len(os.sched_getaffinity(0))}")
except AttributeError:
    print (f"Physical Cores: {multiprocessing.cpu_count()}")
```

```python
inertia = []
k_range = range(1,8)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(xScaled)
    inertia.append(kmeans.inertia_)
```

```python
# plot Elbow curve
plt.figure(figsize=(8,5))
plt.plot(k_range, inertia, marker='o')
plt.title("Elbow Method for optimal K")
plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
```

```python
# The curve starts to flatten between 3 and 4 - the elbow
```
